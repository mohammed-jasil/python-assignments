```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Zara", "Jasil", "Rahul", "Anjali", "Arun"] 
print("age_list:", age_list)
print("name_list:", name_list)
```

```python
name_list.append("Yazhini")
print("a. After append:", name_list)
age_list.insert(2, 30)
print("b. After insert:", age_list)
name_list.remove("Yazhini")
print("c. After remove:", name_list)
age_list.pop()
print("d. After pop:", age_list)
age_list.extend([29, 30, 26])
print("e. After extend:", age_list)
age_list.sort(reverse=True)
print("f. sorted descending:", age_list)
print("g. Max age:", max(age_list))
print(" Min age:", min(age_list))
print(" Sum of age:", sum(age_list))      
```

```python
print("a. First element:", name_list[0])
print("b. Last element:", name_list[-1])
print("c.Index 2 to 4:", name_list[2:5])
print("d. reverse order:", name_list[::-1])
```

```python
student_marks = {
    "Zara": 85,
    "Jasil": 90,
    "Rahul": 75,
    "Anjali": 88,
    "Arun": 92,
}
print("a. Initial dictionary:", student_marks)
print("b. Mark of Jasil:", student_marks["Jasil"])
student_marks["Janani"] = 80
print("c. After adding Janani:", student_marks)
student_marks["Zara"] = 82
print("d. After updating Zara's marks:", student_marks) 
```

```python
print("Keys:", student_marks.keys())
print("values:", student_marks.values())
print("key-value pairs:", student_marks.items())
```

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print("\nc. set1:", set1)
print("   set2:", set2)
print("d. Union:", set1.union(set2))
print("   Intersection:", set1.intersection(set2))
```
